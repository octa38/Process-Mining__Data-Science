```python
import pandas as pd
import matplotlib.pyplot as plt
from pm4py.objects.log.importer.xes import importer as xes_importer
from datetime import datetime
import os
import networkx as nx

# Define a function to calculate duration in seconds between two times
def calculate_duration_in_seconds(start_time, end_time):
    duration = end_time - start_time
    return duration.total_seconds()

# Set the directory path where the log files are located
directory_path = '/home/jovyan/Project/FilteredFiles'

# Initialize dictionary to store global event durations
global_event_durations = {}

# Loop over each file in the directory to calculate global average durations
for filename in os.listdir(directory_path):
    # Check if the file is an XES log file
    if filename.endswith(".xes"):
        # Construct full file path and import the log
        file_path = os.path.join(directory_path, filename)
        log = xes_importer.apply(file_path)

        # Loop through each case in the log
        for case_index, case in enumerate(log):
            # Temporary dictionary to store start times of events
            temp_start_times = {}
            # Loop through each event in the case
            for event_index, event in enumerate(case):
                # Extract event name and lifecycle transition
                event_name = event['concept:name']
                lifecycle = event['lifecycle:transition']
                # Convert timestamp to datetime object
                current_time = datetime.fromtimestamp(event['time:timestamp'].timestamp())

                # If event is starting, record the start time
                if lifecycle == 'Start':
                    temp_start_times[event_name] = current_time
                # If event is completing, calculate duration and record it
                elif lifecycle == 'Complete' and event_name in temp_start_times:
                    start_time = temp_start_times[event_name]
                    duration = calculate_duration_in_seconds(start_time, current_time)

                    # Add duration to global durations list or create a new list
                    if event_name in global_event_durations:
                        global_event_durations[event_name].append(duration)
                    else:
                        global_event_durations[event_name] = [duration]

                    # Remove the start time entry as the event is complete
                    del temp_start_times[event_name]

# Calculate the global average duration for each event
global_average_durations = {event_name: sum(durations) / len(durations) for event_name, durations in global_event_durations.items()}

# Initialize dictionary to store the best scheduling details
best_scheduling = {
    'filename': None,
    'dataframe': None,
    'critical_path': [],
    'critical_duration': float('inf'),
    'graph': None,
    'log': None
}

# Loop over each file again to calculate local average durations and find the best scheduling
for filename in os.listdir(directory_path):
    # Check if the file is an XES log file
    if filename.endswith(".xes"):
        # Construct full file path and import the log
        file_path = os.path.join(directory_path, filename)
        log = xes_importer.apply(file_path)

        # Initialize dictionaries to store event durations and local averages
        event_durations = {}
        local_average_durations = {}

        # Loop through each case in the log
        for case_index, case in enumerate(log):
            # Temporary dictionary to store start times of events
            temp_start_times = {}
            # Loop through each event in the case
            for event_index, event in enumerate(case):
                # Extract event name and lifecycle transition
                event_name = event['concept:name']
                lifecycle = event['lifecycle:transition']
                # Convert timestamp to datetime object
                current_time = datetime.fromtimestamp(event['time:timestamp'].timestamp())

                # If event is starting, record the start time
                if lifecycle == 'Start':
                    temp_start_times[event_name] = current_time
                # If event is completing, calculate duration and record it
                elif lifecycle == 'Complete' and event_name in temp_start_times:
                    start_time = temp_start_times[event_name]
                    duration = calculate_duration_in_seconds(start_time, current_time)

                    # Add duration to event durations list or create a new list
                    if event_name in event_durations:
                        event_durations[event_name].append(duration)
                    else:
                        event_durations[event_name] = [duration]

                    # Remove the start time entry as the event is complete
                    del temp_start_times[event_name]

        # Calculate the local average duration for each event
        for event_name, durations in event_durations.items():
            local_average = sum(durations) / len(durations)
            local_average_durations[event_name] = local_average

        # Create a dataframe with local average durations
        df_local = pd.DataFrame(list(local_average_durations.items()), columns=['Event Name', 'Local Average Duration (Seconds)'])

        # Define the activities and their dependencies
        activities = ['Arrival', 'Welding', 'Painting', 'Drilling', 'Drain', 'Sawing']
        dependencies = {
            'Arrival': ['Drilling', 'Sawing', 'Painting', 'Welding'],
            'Drilling': ['Sawing', 'Painting', 'Welding', 'Drain'],
            'Sawing': ['Drilling', 'Painting', 'Welding', 'Drain'],
            'Painting': ['Drilling', 'Sawing', 'Welding', 'Drain'],
            'Welding': ['Drilling', 'Sawing', 'Painting', 'Drain']}
        durations = local_average_durations

        # Initialize a directed graph and add nodes and edges based on activities and dependencies
        G = nx.DiGraph()
        for activity in activities:
            G.add_node(activity, duration=durations[activity])

        for activity, following_activities in dependencies.items():
            for next_activity in following_activities:
                G.add_edge(activity, next_activity)

        # Define a function to find the critical path in the process flow. It is used to enforce that the path goes through the 4 activities before drain.
        def find_critical_path(graph):
            longest_path = []
            longest_duration = 0
            for path in nx.all_simple_paths(graph, source='Arrival', target='Drain'):
                current_duration = sum(graph.nodes[node]['duration'] for node in path)
                if current_duration > longest_duration:
                    longest_duration = current_duration
                    longest_path = path
            return longest_path, longest_duration

        # Find the critical path and duration
        critical_path, critical_duration = find_critical_path(G)

        # Update the best scheduling if the current one is better
        if critical_duration < best_scheduling['critical_duration']:
            best_scheduling['filename'] = filename
            best_scheduling['dataframe'] = df_local.copy() 
            best_scheduling['critical_path'] = critical_path
            best_scheduling['critical_duration'] = critical_duration
            best_scheduling['graph'] = G
            best_scheduling['log'] = log

# Print and plot the details of the best scheduling
if best_scheduling['filename']:
    # Print basic details of the best scheduling
    print("\nDetails of the dataset with the shortest over-all time:")
    print(f"Filename: {best_scheduling['filename']}")
    print(f"Path: {best_scheduling['critical_path']}")
    print(f"Duration: {best_scheduling['critical_duration']}")

    # Print the local average duration dataframe
    print("\nDataframe:")
    print(best_scheduling['dataframe'])

    # Plot the process flow graph
    plt.figure(figsize=(12, 8))
    pos = nx.spring_layout(best_scheduling['graph'])
    nx.draw(best_scheduling['graph'], pos, with_labels=True, node_size=5000, node_color="skyblue")
    plt.title("Process Flow")
    plt.show()

    # Plot a comparison of local and global average durations
    events = best_scheduling['dataframe']['Event Name'].tolist()
    local_durations = best_scheduling['dataframe']['Local Average Duration (Seconds)'].tolist()
    global_durations = [global_average_durations[event] for event in events]

    x = range(len(events))
    plt.figure(figsize=(10, 6))
    bar_width = 0.35
    plt.bar(x, local_durations, width=bar_width, label='Local', color='skyblue')
    plt.bar([p + bar_width for p in x], global_durations, width=bar_width, label='Global', color='orange')

    plt.xlabel('Event Name')
    plt.ylabel('Average Duration (Seconds)')
    plt.title(f'Event Durations for {best_scheduling["filename"]}')
    plt.xticks([p + 0.5 * bar_width for p in x], events, rotation=90)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # Identify the most likely bottleneck event by comparing local and global averages
    highest_duration_difference = 0
    most_likely_bottleneck = ""

    for event_index, event_name in enumerate(events):
        local_average = local_durations[event_index]
        global_average = global_average_durations.get(event_name, 0)
        print(f"Event: {event_name}, Local Average: {local_average}, Global Average: {global_average}")

        if local_average > global_average:
            duration_difference = local_average - global_average
            print(f"Potential Bottleneck: {event_name}, Difference: {duration_difference}")
            if duration_difference > highest_duration_difference and event_name != "Transport":
                highest_duration_difference = duration_difference
                most_likely_bottleneck = event_name
```

```python
welding = local_average_durations.get('Welding')
painting = local_average_durations.get('Painting')
drilling = local_average_durations.get('Drilling')
sawing = local_average_durations.get('Sawing')

jobs = [
    [(0, drilling), (1, sawing), (2, painting), (3, welding)],
    [(1, sawing), (0, drilling), (2, painting), (3, welding)],
    [(3, welding), (2, painting),  (0, drilling), (1, sawing)],
    [(2, painting), (3, welding), (0, drilling), (1, sawing)], 
]

def global_schedule_jobs(jobs):
    scheduling_data = []
    # pointer to current operation for each job
    job_pointers = [0] * len(jobs)  
    max_workstation = max(op[0] for job in jobs for op in job)
    machine_availability = {machine_number: 0 for machine_number in range(max_workstation + 1)}
    job_availability = [0] * len(jobs)
    
    #looping through the jobs until all operations are scheduled
    while any(pointer < len(jobs[0]) for pointer in job_pointers):
        #finding the next operation to be scheduled
        next_op_idx, next_job_idx = None, None
        next_start_time = float('inf')
        
        #calculating a job's possible start and choosing the max from when it is available and when the workstation is available
        for j, pointer in enumerate(job_pointers):
            if pointer < len(jobs[0]):
                workstation, time = jobs[j][pointer]
                start_time = max(job_availability[j], machine_availability[workstation])
                if start_time < next_start_time:
                    next_start_time = start_time
                    next_op_idx = pointer
                    next_job_idx = j
        
        #scheduling the next operation and computing its end time
        workstation, time = jobs[next_job_idx][next_op_idx]
        end_time = next_start_time + time
        
        #updating machine and job availability
        machine_availability[workstation] = end_time
        job_availability[next_job_idx] = end_time
        
        #storing the data for scheduling
        while len(scheduling_data) <= next_job_idx:
            scheduling_data.append([])
        scheduling_data[next_job_idx].append({
            'workstation': workstation,
            'time': time,
            'start_time': next_start_time,
            'end_time': end_time
        })
        
        #moving to the next operation of this job
        job_pointers[next_job_idx] += 1
    
    return scheduling_data

#computing global schedule
scheduling_data = global_schedule_jobs(jobs)

#showing the results
for i, job_schedule in enumerate(scheduling_data):
    print('job', i + 1, ":")
    for op in job_schedule:
        print(f"Machine {op['workstation']}: Start Time {op['start_time']}, End Time {op['end_time']}")
```

```python
#setting the durations of each action
average_durations = {
    'Arrival': 0.0, 
    'Welding': 353.43, 
    'Painting': 298.45, 
    'Drilling': 91.54, 
    'Drain': 0.0, 
    'Sawing': 97.95
}

# Average transport duration
transport_duration = 60

# Actions that still need to happen
actions = ["Arrival", "Welding", "Painting", "Drilling", "Sawing", "Drain"]

# Predicted end time of process
predicted_duration = average_durations['Arrival'] + 5 * transport_duration + average_durations['Drain'] + average_durations['Welding'] + average_durations['Drilling'] + average_durations['Sawing'] + average_durations['Painting']

while True:
    #inputting the activity, start_time and end_time. Each have to be inputted in a correct form, we didn't account for mistakes.
    activity = input('Enter the activity: ')
    start_time = input('Enter the start time (hh:mm:ss): ')
    end_time = input('Enter the end time (hh:mm:ss): ')
    
    #getting a numerical value of start time and end time in seconds
    start_time_list = start_time.split(":")
    end_time_list = end_time.split(":")
    start_time_num = int(start_time_list[0]) * 3600 + int(start_time_list[1]) * 60 + int(start_time_list[2])
    end_time_num = int(end_time_list[0]) * 3600 + int(end_time_list[1]) * 60 + int(end_time_list[2])
    
    # How much time did this action take
    actual_duration = end_time_num - start_time_num
    #removing inputted activity from the list as it happened
    actions.remove(activity)
    
    if activity != "Drain":
        #if live duration is longer than average, calculate difference
        if actual_duration > average_durations[activity]:
            difference = round(actual_duration - average_durations[activity], 2)
            print(f"It took you {difference} seconds longer than expected to complete {activity}.")
        #if live duration is shorter than average, calculate difference
        elif actual_duration < average_durations[activity]:
            difference = round(average_durations[activity] - actual_duration, 2)
            print(f"It took you {difference} seconds shorter than expected to complete {activity}.")
        #they took the same amount of time
        else:
            print(f"It took you exactly {actual_duration} seconds which is the same as the average duration to complete {activity}.")
        print("Following actions still need to be completed: ")
        print(*actions, sep = ", ")
        predicted_duration -= average_durations[activity]
        predicted_duration += actual_duration
        # Calculate the expected end time
        hours = round(predicted_duration // 3600)
        minutes = round((predicted_duration % 3600) // 60)
        seconds = round(predicted_duration % 60)
        # Calculate the total seconds
        total_seconds = seconds + int(start_time_list[2])

        # Carry-over calculations
        carry_minutes = total_seconds // 60  # Number of minutes to carry over
        total_seconds %= 60  # Remaining seconds

        total_minutes = int(start_time_list[1]) + minutes + carry_minutes  # Add carried minutes to start_minutes

        carry_hours = total_minutes // 60  # Number of hours to carry over
        total_minutes %= 60  # Remaining minutes

        total_hours = int(start_time_list[0]) + hours + carry_hours  # Add carried hours to start_hours
        if activity != "Arrival":
            predicted_duration = predicted_duration - actual_duration - transport_duration
        else:
            predicted_duration = predicted_duration - actual_duration
        print(f"Based on that the end time of the process is now {str(total_hours)}:{str(total_minutes)}:{str(total_seconds)}")

    else:
        print("We are done!")
        break
```
